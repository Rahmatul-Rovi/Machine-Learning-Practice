```python
import pandas as pd
```

```python
sales_data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "Product": ["Laptop", "Shirt", "Book", "Mobile", "Shoes", "Pen", "Headphones", "Notebook"],
    "Category": ["Electronics", "Clothing", "Stationery", "Electronics", "Clothing", "Stationery", "Electronics", "Stationery"],
    "Quantity": [2, 5, 3, None, 2, 10, 4, 6],
    "Price": [800, 25, 15, 500, 60, None, 120, 5],
    "Region": ["North", "South", "East", "West", "North", "South", "East", "West"],
    "Discount": [None, 0.10, None, 0.05, 0.15, 0.05, 0.20, None]
}
```

```python
print(sales_data)
```

```python
# Convert dictionary into DataFrame.
df = pd.DataFrame(sales_data)
print(df)
```

```python
# Print the first 5 rows of the dataset
print(df.iloc[0:5])
```

```python
# Filter orders where the Price is greater than 100.
greater_than=df[df["Price"]>100]
print(greater_than)
```

```python
# Display all sales records where the Region is "North".
print(df.iloc[5:5])
```

```python
# Extract the value of the 5th record’s price.
print(df.iloc[4,4])
```

```python
# Find all products sold in the "Electronics" category.
print(df["Category"])
```

```python
# Print the count of total missing values in the dataframe.
print(df.isna().sum())
```

```python
# Drop all rows that contain any missing value without modifying the original dataframe.
r = df.drop(columns=["Quantity",  "Price",   "Discount"],inplace=True)
print(r)
```

```python
# Replace missing values in the Quantity column with the average quantity.
df["Discount"]= df["Discount"].fillna(0)
print(df)
```

```python
# Replace missing values in the Discount column with 0 (no discount).
df['Price'] = df['Price'].fillna(df['Price'].min())
print(df)
```

```python
#  Find the average price of all products in the dataset.
mean_r = df['Price'].mean()
print(mean_r)
```
