```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Student data
data = {
    "Name": ["Rahim", "Karim", "Jamal", "Raju", "Sohel", "Mitu", "Rima"],
    "Math": [85, 92, 78, 88, 76, 95, 70],
    "English": [80, 85, 72, 90, 82, 88, 75],
    "Science": [90, 88, 80, 85, 78, 92, 83]
}
```

```python
df = pd.DataFrame(data)
```

```python
# Average calculate
df["Average"] = (df["Math"] + df["English"] + df["Science"]) / 3
```

```python
# Grade dewa
def get_grade(avg):
  if avg>=90: return "A+"
  elif avg>=80: return "A"
  elif avg >= 70: return "B"
  else: return "C"
```

```python
df["Grade"] = df["Average"].apply(get_grade)
```

```python
print(df)
```

```python
print(f"\nClass Average: {df['Average'].mean():.2f}")
```

```python
print(f"Best Student: {df.loc[df['Average'].idxmax(), 'Name']}")
```

```python
# Chart
plt.figure(figsize=(10, 6))
plt.bar(df["Name"], df["Average"], color="purple")
plt.title("Student Performance")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")
plt.axhline(y=df["Average"].mean(), color="red",
            linestyle="--", label="Class Average")
plt.legend()
plt.show()
```
