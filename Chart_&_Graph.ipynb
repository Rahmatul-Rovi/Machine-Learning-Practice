```python
import matplotlib.pyplot as plt

# Simple bar chart
subjects = ["Bangla", "English", "Mathematics", "Science", "ICT"]
marks = [85, 90, 80, 75, 82]

plt.figure(figsize=(10,6))
plt.bar(subjects, marks, color="blue")
```

```python
plt.title("Subject Wise Marks")
```

```python
plt.xlabel("Subject")
```

```python
plt.ylabel("Marks")
```

```python
plt.show()
```
