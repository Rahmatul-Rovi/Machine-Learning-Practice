```python
import matplotlib.pyplot as plt

# Simple bar chart
subjects = ["Math", "English", "Science", "Bengali", "ICT"]
marks = [85, 78, 92, 88, 76]

plt.figure(figsize=(10, 6))
plt.bar(subjects, marks, color="blue")
plt.title("Subject wise Marks")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()

# Line chart
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [1000, 1200, 900, 1500, 1300, 1800]

plt.figure(figsize=(10, 6))
plt.plot(months, sales, marker="o", color="green", linewidth=2)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (BDT)")
plt.grid(True)
plt.show()
```
