```python
import pandas as pd
# DataFrame
data = {
    "Name": ["Rahim", "Karim", "Jamal", "Raju", "Sohel"],
    "Age": [20, 22, 21, 23, 20],
    "Marks": [85, 92, 78, 88, 76],
    "City": ["Dhaka", "Chittagong", "Dhaka", "Sylhet", "Dhaka"]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
```

```python
print(df["Name"])
print(df["Marks"])
```

```python
dhaka_students = df[df["City"] == "Dhaka"]
print(dhaka_students)
```

```python
good_students = df[df["Marks"] > 80]
print(good_students)
```

```python
print(f"Average marks: {df['Marks'].mean():.2f}")
print(f"Highest marks: {df['Marks'].max()}")
```
