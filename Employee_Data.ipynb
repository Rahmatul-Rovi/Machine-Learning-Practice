```python
employee_data = {
    "emp_ID": [101, 102, 103, 104, 105],
    "Post": ["Manager", "Developer", "Designer", "Analyst", "Intern"],
    "SalaryPerHour": [50, None, 35, 38, 15],
    "Bonus": [500, 400, None, 350, 100],
    "WorkingHour": [160, 170, 150, None, 30],
}
print(employee_data)
```

```python
import pandas
data= pandas.DataFrame(employee_data)
print(data)
```

```python
print(data.isna())
```

```python
print(data.isna())
```
