```python
#For Loop
for i in range (5):
  print(f"Number: {i}")

#List Loop
fruits = ["Apple", "Banana", "Mango"]
for fruit in fruits:
  print(f"Fruits: {fruit}")

#While Loop
count = 0
while count < 5:
  print(f"Count: {count}")
  count +=1

#Real Example
marks = [80,70, 90, 75, 85]
total = 0
for mark in marks:
  total += mark
  average = total /len(marks)
  print(f"Average Marks: {average}")
```
