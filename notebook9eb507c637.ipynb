```python
min_max_data = (data - data.min()) / (data.max() - data.min())
print(min_max_data)
```

```python
#Z-Scope-Scaling
zscope_data = (data - data.mean()) / data.std()
print(zscope_data)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)
print(scaled_data)
scaled_data = pandas.DataFrame(scaled_data , columns = data.columns)
print(scaled_data)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
print(scaled_data)
scaled_data = pandas.DataFrame(scaled_data , columns = data.columns)
print(scaled_data)
```
