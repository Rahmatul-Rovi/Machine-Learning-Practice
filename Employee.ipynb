```python
employee_data = {
    "emp_ID": ['E101', 'E102', 'E103', 'E104', 'E105'],
    "Post": ["Manager", "Engineer", "Analyst", "Engineer", "Manager"],
    "WorkingHour": ['Full-Time', 'Part-Time', 'Full-Time', 'Full-Time', 'Part-Time'],
    "Bonus": ['High', 'Medium', 'Low', 'Medium', 'High'],
    "Salary": [80000, 60000, 50000, 65000, 85000],
}
print(employee_data)
```

```python
import pandas
data = pandas.DataFrame(employee_data)
print(data)
```

```python
from sklearn.preprocessing import LabelEncoder
label=LabelEncoder()

categorical_columns = ['emp_ID', "Post", "WorkingHour", "Bonus"]

for col in categorical_columns :
    data[col] = label.fit_transform(data[col])
    print('\nEncoded DataFrame:')
    print(data)
```

```python
encode = pandas.get_dummies(
    data,
    columns = ['Post', 'WorkingHour', 'Bonus'],
    dtype=int
)
print(encode)
```
